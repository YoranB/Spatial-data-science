```python
import pandas as pd
import glob
import os

# Pad naar je map met bestanden
path = "../../data/raw/crime_monthly"

# Zoek alle CSV bestanden in de map
all_misdrijf_files = sorted(glob.glob(os.path.join(path, "*.csv")))

print("Gevonden bestanden:")
for f in all_misdrijf_files:
    print(" -", f)

print("Totaal aantal bestanden:", len(all_misdrijf_files))

# Controle: zijn er wel bestanden gevonden?
if len(all_misdrijf_files) == 0:
    raise ValueError("Er zijn geen CSV bestanden gevonden in de opgegeven map.")

# Inlezen en samenvoegen
dfs = []

for file in all_misdrijf_files:
    print(f"Inlezen: {file}")
    df = pd.read_csv(file, sep=None, engine="python")
    dfs.append(df)

# Controle: zijn er dataframes ingelezen?
if len(dfs) == 0:
    raise ValueError("Er zijn geen dataframes om samen te voegen.")

df_all = pd.concat(dfs, ignore_index=True)

df_all
```

```python
df_crime_average_per_year = df_all.groupby('\ufeffWijk')['Totaal misdrijven'].mean().reset_index() 
df_crime_average_per_year.rename(columns={'\ufeffWijk':'Wijk'}, inplace=True) 
df_crime_average_per_year['Totaal misdrijven'] = df_crime_average_per_year['Totaal misdrijven'].round(0)  
df_crime_average_per_year
```

```python
df_crime_average_per_year.to_csv("../../data/intermediate/df_crime_average_per_year.csv", index=False)
```
