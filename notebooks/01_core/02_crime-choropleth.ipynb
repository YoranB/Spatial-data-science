```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# Geopackage data
gemeenten = gpd.read_file("../../data/raw/boundaries/wijken_en_gemeenten.gpkg", layer="gemeenten")
amsterdam_grens = gemeenten[gemeenten['gemeentenaam'] == 'Amsterdam']

wijken = gpd.read_file("../../data/raw/boundaries/wijken_en_gemeenten.gpkg", layer="wijken")
amsterdam_wijken = gpd.clip(wijken, amsterdam_grens)

total = pd.read_csv("../../data/final/df_final_model.csv")

# KOPPELEN
amsterdam_data = amsterdam_wijken.merge(
    total,
    left_on='wijknaam',
    right_on='neighbourhood_name',
    how='left'
)

# PLOTTEN
fig, ax = plt.subplots(figsize=(15, 12), dpi=300)

amsterdam_data.plot(
    column='crime_rate_per_1000',
    scheme='quantiles',
    k=5,
    cmap='OrRd',
    linewidth=0.5,
    edgecolor='grey',
    legend=True,
    legend_kwds={'loc': 'upper left', 'title': "Misdrijven per 1000 inw."},
    missing_kwds={'color': 'lightgrey', 'label': 'Geen data'},
    ax=ax
)

ax.set_title("Criminaliteit in Amsterdam (Per 1000 inwoners)", fontsize=18)
ax.set_axis_off()

plt.show()
```
