```python
import pandas as pd
```

```python
df_lights = pd.read_csv("data/LICHTPUNTEN_met_wijk.csv", sep=";") 
df_crimes = pd.read_csv("data/df_crime_average_per_year.csv") 
df_distances = pd.read_csv("data/datapoging2.csv", sep=";")
df_distances.head()
```

```python
#this was used to explore the differences in wijk names between the two datasets

set1 = set(df_lights['wijknaam'].dropna().str.strip().str.lower())
set2 = set(df_crimes['Wijk'].dropna().str.strip().str.lower()) 

in_both = sorted(set1 & set2)

# Alleen in df1
only_set1 = sorted(set1 - set2)

# Alleen in df2
only_set2 = sorted(set2 - set1)
print(only_set2)
print(only_set1)
```

```python
#this was used to explore the differences in wijk names between the two datasets

set1 = set(df_lights['wijkcode'].dropna().str.strip().str.lower())
set2 = set(df_distances['WijkenEnBuurten'].dropna().str.strip().str.lower()) 

in_both = sorted(set1 & set2)

# Alleen in df1
only_set1 = sorted(set1 - set2)

# Alleen in df2
only_set2 = sorted(set2 - set1)
print(in_both)
```

```python
#lantaarn data klaarmaken om te mergen 
df_lights_per_wijk = (
    df_lights
    .groupby(['wijkcode', 'wijknaam'])
    .size()
    .reset_index(name='aantal_lantaarns')
)

df_lights_per_wijk.head()
```

```python
#de wijknaam colomnen schoon maken van beide data sets zodat ik daar op kan mergen 
df_lights_per_wijk['wijknaam_clean'] = (
    df_lights_per_wijk['wijknaam']
    .str.strip()
    .str.lower()
)

df_crimes['wijknaam_clean'] = (
    df_crimes['Wijk']
    .str.strip()
    .str.lower()
)

df_merged = df_lights_per_wijk.merge(
    df_crimes, 
    on='wijknaam_clean',
    how='inner'
)
#merging the lantaarns data with crime data
df_merged = df_merged.drop(columns=['wijknaam_clean', 'Wijk'])  
df_merged.head()
```

```python
df_merged['wijkcode_clean']= (
    df_merged['wijkcode'] 
    .str.strip()
    .str.lower()
)

df_distances['wijkcode_clean'] =  (
    df_distances['WijkenEnBuurten']
    .str.strip()
    .str.lower()
)


df_merged_total = df_merged.merge(
    df_distances,  
    on='wijkcode_clean', 
    how='inner'   # alleen wijken die in beide voorkomen
) 

df_merged_total.columns
```

```python
df_merged_total= df_merged_total.drop(columns=['WijkenEnBuurten', 'Gemeentenaam_1', 'Codering_3', 'Gemeentenaam_1', 'ID', 'SoortRegio_2', 'IndelingswijzigingWijkenEnBuurten_4'])
```

```python
cols_to_drop = [col for col in df_merged_total.columns if 'Binnen' in col]

df_merged_total = df_merged_total.drop(columns=cols_to_drop) 

df_merged_total 
```

```python
df_merged_total.to_csv("data/total_table.csv",  index=False)
```
